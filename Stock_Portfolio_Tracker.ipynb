```python
stock_prices = {
    "AAPL": 180,
    "TSLA": 250,
    "GOOGL": 170,
    "MSFT": 420,
    "AMZN": 200
}


def portfolio_tracker():
    print("=" * 60)
    print("            Stock Portfolio Tracker")
    print("=" * 60)

    total_investment = 0
    portfolio = []

    while True:
        print("\nAvailable Stocks:")
        for stock, price in stock_prices.items():
            print(f"{stock} : ${price}")

        stock_name = input("\nEnter Stock Name: ").upper()

        if stock_name not in stock_prices:
            print("Invalid stock name. Please try again.")
            continue

        quantity = int(input("Enter Quantity: "))

        investment = stock_prices[stock_name] * quantity
        total_investment += investment

        portfolio.append(
            (stock_name, quantity, stock_prices[stock_name], investment)
        )

        choice = input("\nDo you want to add another stock? (yes/no): ").lower()

        if choice != "yes":
            break

    print("\n" + "=" * 60)
    print("Portfolio Summary")
    print("=" * 60)

    for stock, qty, price, value in portfolio:
        print(f"{stock:<8} Quantity: {qty:<3} Price: ${price:<5} Investment: ${value}")

    print("-" * 60)
    print(f"Total Investment Value: ${total_investment}")

    # Save report
    with open("portfolio_report.txt", "w") as file:
        file.write("Stock Portfolio Report\n")
        file.write("=" * 40 + "\n")

        for stock, qty, price, value in portfolio:
            file.write(
                f"{stock} | Quantity: {qty} | Price: ${price} | Investment: ${value}\n"
            )

        file.write("-" * 40 + "\n")
        file.write(f"Total Investment Value: ${total_investment}")

    print("\nPortfolio report saved as 'portfolio_report.txt'.")


if __name__ == "__main__":
    portfolio_tracker()
```
